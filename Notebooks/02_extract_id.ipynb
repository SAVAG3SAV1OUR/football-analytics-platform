```python
import requests
import json
import os
from dotenv import load_dotenv
```

```python
load_dotenv()

API_KEY = os.getenv("API_KEY")
headers = {'x-api-key': API_KEY}
BASE_URL = os.getenv("BASE_URL")
```

## Get Tournament IDs

```python
leagues = requests.get(
    f"{BASE_URL}/leagues",
    headers=headers
).json()

#print(json.dumps(leagues, indent=4))
```

```python
TARGET_COMPS = [
    "premier-league",
    "uefa-champions-league",
    'laliga',
    'serie-a',
    'bundesliga'
]
```

```python
competitions = []

for country in leagues.get("countries",[]):
    for league in country.get("leagues", []):
        competitions.append(league)

#for league in leagues.get("leagues", []):
    #competitions.append(league)
```

```python
norm_comps = []

for league in competitions:
    norm_comps.append({
        "comp_id": league["id"],
        "comp_name": league["name"],
        "slug": league["slug"],
        "country": league.get("countryName"),
        "category": league.get("categoryId")
    })

#selected_comps = [
 #   competition
 #   for competition in norm_comps
 #   if competition['slug'] in TARGET_COMPS
#]
```
