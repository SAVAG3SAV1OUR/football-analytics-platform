```python
import os
from dotenv import load_dotenv
import requests
import json

load_dotenv()

API_KEY = os.getenv("API_KEY")
BASE_URL = os.getenv("BASE_URL")
headers = {"x-api-key": API_KEY}
```

```python
season = requests.get(
    f"{BASE_URL}/tournaments/17/seasons",
    headers=headers
).json()

print(json.dumps(season, indent=4))
```

```python
season_info = requests.get(
    f"{BASE_URL}/tournament/17/season/29415/info",
    headers=headers
).json()

print(json.dumps(season_info, indent=4))
```

```python
#Season Teams
season_teams = requests.get(
    f"{BASE_URL}/tournament/7/season/76953/round/1",
    headers=headers
).json()

print(json.dumps(season_teams, indent=4))
```

```python
season_teams = requests.get(
    f"{BASE_URL}/tournament/7/season/76953/knockout",
    headers=headers
).json()

print(json.dumps(season_teams, indent=4))
```

```python
seasons = requests.get(f"{BASE_URL}/api/tournaments/17/seasons", headers=headers).json()["seasons"]
current = seasons[0]
matches = requests.get(
    f"{BASE_URL}/api/tournament/17/season/{current['id']}/events/last/0",
    headers=headers,
).json()

print(json.dumps(matches, indent=4))
```
